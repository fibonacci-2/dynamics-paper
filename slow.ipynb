```python
import pandas as pd
import requests


df = pd.read_csv('data/mindset/all-posts/final_ocd.csv')

def fetch_post(ids):
    print(f"Fetching posts for IDs: {ids}")
    posts_url = f'https://arctic-shift.photon-reddit.com/api/posts/ids?ids={ids}'


    posts = {}
    try:
        # Fetch posts
        posts_response = requests.get(posts_url, verify=False)  # Disable SSL verification
        posts_response.raise_for_status()  # Raise an exception for HTTP errors
        posts = posts_response.json().get('data', [])  # Get the list of posts

   
    except Exception as e:
        print(f"Failed to fetch data for user Error: {e}")

    return posts

# append post created_utc using the function if post id length is 6
posts = fetch_post(','.join(df[df['user_id'] == 217572]['post_id'].filter(lambda x: len(str(x)) == 6).astype(str).tolist()))
posts, df.head()
```
